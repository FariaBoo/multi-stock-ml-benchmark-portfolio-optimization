```python
import pandas as pd
from datetime import datetime, timedelta
import time
import logging
import zipfile
import os
from IPython.display import display, FileLink
 
# Set up logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)
 
# Define the tickers for your 7 portfolio stocks
tickers = ['MSFT', 'LLY', 'JPM', 'AMZN', 'PG', 'XOM', 'NEE']
 
# Set the time period (last 30 days)
end_date = datetime.now()
start_date = end_date - timedelta(days=30)
 
# Configuration for download retries and rate limiting
max_retries = 5
base_delay = 10  # seconds
chunk_delay = 5  # seconds between chunks
 
# Loop through each stock ticker
for ticker in tickers:
    logger.info(f"=== Starting data collection for {ticker} ===")
 
    all_data = pd.DataFrame()
    current_end = end_date
 
    while current_end > start_date:
        current_start = max(current_end - timedelta(days=7), start_date)
        logger.info(f"Fetching {ticker} data from {current_start} to {current_end}")
 
        for attempt in range(max_retries):
            try:
                chunk = yf.download(ticker, start=current_start, end=current_end, interval="1m", auto_adjust=False)
 
                if not chunk.empty:
                    all_data = pd.concat([all_data, chunk])
                    logger.info(f"Successfully fetched data for {ticker} ({current_start} to {current_end})")
                    break
                else:
                    logger.warning(f"Empty data for {ticker} ({current_start} to {current_end}), retrying...")
                    if attempt < max_retries - 1:
                        delay = base_delay * (2 ** attempt)
                        logger.info(f"Waiting {delay} seconds before retrying...")
                        time.sleep(delay)
                    else:
                        logger.error(f"Max retries reached for {ticker} ({current_start} to {current_end}), skipping...")
 
            except Exception as e:
                logger.error(f"Error on attempt {attempt + 1}/{max_retries} for {ticker}: {e}")
                if attempt < max_retries - 1:
                    delay = base_delay * (2 ** attempt)
                    logger.info(f"Waiting {delay} seconds before retrying...")
                    time.sleep(delay)
                else:
                    logger.error(f"Max retries reached for {ticker} ({current_start} to {current_end}), skipping...")
 
        time.sleep(chunk_delay)
        current_end = current_start
 
    if not all_data.empty:
        all_data = all_data[~all_data.index.duplicated(keep='first')]
        all_data = all_data.sort_index()
        file_name = f"{ticker}.csv"
        all_data.to_csv(file_name)
        logger.info(f"Data saved to {file_name}")
        logger.info(f"First few rows of data for {ticker}:")
        print(all_data.head())
    else:
        logger.warning(f"No data collected for {ticker}")
```

```python
# Step 2: Load and Clean Data
print("\nStep 2: Loading and cleaning data...")

# Use the same TARGET_COLUMNS list defined in Step 1, instead of redefining it
# here as 'tickers' — avoids two separate lists drifting out of sync later.
tickers = TARGET_COLUMNS
stock_data = {}
news_data = None

# Attempt to load optional news_data.csv if it exists in local directory
if os.path.exists('news_data.csv'):
    try:
        news_data = pd.read_csv('news_data.csv')
        print(f"News data loaded: {len(news_data)} rows")
        print("News data columns:", news_data.columns.tolist())
        print("News data sample:\n", news_data.head(2))
    except Exception as e:
        print(f"Error loading news data: {e}")
        news_data = None


def load_yfinance_csv(file_name):
    """
    FIX: yfinance's to_csv() writes 3 header rows when columns are a
    MultiIndex (Price level, Ticker level, and a row for the index name) —
    e.g.:
        Price,Adj Close,Close,High,Low,Open,Volume
        Ticker,MSFT,MSFT,MSFT,MSFT,MSFT,MSFT
        Datetime,,,,,,
        2026-08-03 09:30:00-04:00,478.31,478.31,479.95,475.17,476.39,3776191
    A plain pd.read_csv() only treats row 1 as headers, so the 'Ticker' and
    'Datetime' rows get read in as DATA — corrupting every price column to
    string/object dtype and breaking .interpolate() downstream.
    This reads it properly: row 1 as headers, rows 2-3 dropped as junk,
    first column renamed to Timestamp, and price columns cast to numeric.
    """
    raw = pd.read_csv(file_name)
    # Drop the 'Ticker' labels row and the blank 'Datetime' row
    raw = raw.iloc[2:].reset_index(drop=True)
    # First column holds the actual datetime strings
    raw = raw.rename(columns={raw.columns[0]: 'Timestamp'})
    # Force all price/volume columns to numeric (they're strings right now)
    for col in raw.columns[1:]:
        raw[col] = pd.to_numeric(raw[col], errors='coerce')
    return raw


# Load and clean CSV files for each of the 7 stocks
for ticker in tickers:
    file_name = f"{ticker}.csv"
    print(f"\nProcessing file: {file_name}")

    if os.path.exists(file_name):
        try:
            df = load_yfinance_csv(file_name)
            print(f"Columns for {ticker}: {df.columns.tolist()}, Initial Rows: {len(df)}")

            # Identify price column — 'Close' is the standard choice
            target_col = ticker if ticker in df.columns else ('Close' if 'Close' in df.columns else 'Adj Close')

            # Interpolate and handle missing price values (now safe — numeric dtype)
            df[target_col] = df[target_col].interpolate().ffill().bfill()
            df = df.dropna(subset=[target_col])

            # Format timestamp column
            df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')
            df = df.dropna(subset=['Timestamp'])

            # Store cleaned data frame in stock_data dictionary
            stock_data[ticker] = df
            print(f"Rows after cleaning for {ticker}: {len(stock_data[ticker])}")
            print(f"Price data sample for {ticker}:\n", stock_data[ticker].head(2))
        except Exception as e:
            print(f"Error loading price data for {ticker}: {e}")
    else:
        print(f"Warning: File {file_name} not found in working directory.")
```

```python
# Step 2.5: Fetch News from FinViz for all 7 stocks
if news_data is None and TRANSFORMERS_AVAILABLE:
    print("\nStep 2.5: No news data uploaded. Fetching news from FinViz...")

    def fetch_finviz_news(ticker):
        url = f'https://finviz.com/quote.ashx?t={ticker}'
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
        }
        try:
            response = requests.get(url, headers=headers, timeout=10)
            response.raise_for_status()
            soup = BeautifulSoup(response.text, 'html.parser')
            news_table = soup.find('table', {'id': 'news-table'})
            if not news_table:
                print(f"No news table found on FinViz for {ticker}")
                return None
            
            extracted_data = []
            for row in news_table.findAll('tr'):
                cells = row.findAll('td')
                if len(cells) < 2:
                    continue
                headline_link = cells[1].find('a')
                if not headline_link:
                    continue
                headline = headline_link.text.strip()
                link = headline_link['href']
                if link.startswith('/'):
                    link = f'https://finviz.com{link}'
                extracted_data.append({'Ticker': ticker, 'Headline': headline, 'Headline_Link': link})
            
            df = pd.DataFrame(extracted_data).head(20)  # Top 20 headlines per stock for speed
            return df
        except Exception as e:
            print(f"Error fetching FinViz data for {ticker}: {e}")
            return None

    def fetch_article_content(url, headline):
        headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
        try:
            response = requests.get(url, headers=headers, timeout=5)
            response.raise_for_status()
            soup = BeautifulSoup(response.text, 'html.parser')
            paragraphs = soup.find_all('p')
            article_text = ' '.join([p.get_text().strip() for p in paragraphs if p.get_text().strip()])
            return article_text if len(article_text.strip()) > 50 else headline
        except Exception:
            return headline

    # Fetch news for all 7 tickers
    all_news = []
    for ticker in tickers:
        print(f"Fetching headlines for {ticker}...")
        df_news = fetch_finviz_news(ticker)
        if df_news is not None:
            all_news.append(df_news)

    if all_news:
        news_data = pd.concat(all_news, ignore_index=True)
        news_data['News_Index'] = range(1, len(news_data) + 1)
        print(f"\nFetched {len(news_data)} total news articles across all tickers.")
        
        print("Extracting article text (this will take ~30 seconds)...")
        news_data['News_detail'] = news_data.apply(
            lambda row: fetch_article_content(row['Headline_Link'], row['Headline']), axis=1
        )
        print("News gathering complete!")
    else:
        print("Failed to fetch news from FinViz. Continuing without news data.")
        news_data = None
```

```python
# Step 2.6: Process News Data with FinBERT for all 7 stocks
if TRANSFORMERS_AVAILABLE and news_data is not None and not news_data.empty:
    print("\nStep 2.6: Processing news data with FinBERT...")
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    
    try:
        torch.cuda.empty_cache()
        tokenizer = AutoTokenizer.from_pretrained("ProsusAI/finbert")
        model = AutoModelForSequenceClassification.from_pretrained("ProsusAI/finbert")
        model.to(device)
        print(f"FinBERT model loaded successfully on {device}.")
    except Exception as e:
        print(f"Failed to load FinBERT model: {e}. Proceeding without sentiment analysis.")
        TRANSFORMERS_AVAILABLE = False

    if TRANSFORMERS_AVAILABLE:
        # Use News_detail or fall back to Headline
        news_data['Text_To_Analyze'] = news_data['News_detail'].fillna(news_data['Headline'])

        def get_finbert_sentiment_batch(texts):
            """Process text in GPU batches for high speed."""
            results = []
            batch_size = 16
            for i in range(0, len(texts), batch_size):
                batch_texts = [str(t)[:512] for t in texts[i:i+batch_size]]
                try:
                    inputs = tokenizer(batch_texts, return_tensors="pt", truncation=True, padding=True, max_length=512).to(device)
                    with torch.no_grad():
                        outputs = model(**inputs)
                    probs = torch.nn.functional.softmax(outputs.logits, dim=-1)
                    results.extend(probs.cpu().numpy())
                except Exception:
                    results.extend([[0.33, 0.33, 0.33]] * len(batch_texts))
            return np.array(results)

        print(f"Analyzing sentiment for {len(news_data)} news items...")
        sentiment_scores = get_finbert_sentiment_batch(news_data['Text_To_Analyze'].tolist())

        # FinBERT outputs logits in order: [Positive, Negative, Neutral]
        news_data[['positive', 'negative', 'neutral']] = sentiment_scores
        news_data['polarity'] = news_data['positive'] - news_data['negative']
        news_data['strength'] = news_data[['positive', 'negative', 'neutral']].max(axis=1)
        news_data['weight'] = news_data['News_Index'] / news_data['News_Index'].max() if news_data['News_Index'].max() > 0 else 1.0
        news_data['impact'] = news_data['polarity'] * news_data['strength'] * news_data['weight']

        print("Sentiment processing complete! Sample scores:\n", news_data[['Ticker', 'positive', 'negative', 'neutral', 'impact']].head(3))

        # Align and merge sentiment features into EACH of the 7 stock dataframes
        sentiment_cols = ['positive', 'negative', 'neutral', 'polarity', 'strength', 'impact']
        
        for ticker in TARGET_COLUMNS:
            if ticker in stock_data and not stock_data[ticker].empty:
                stock_df = stock_data[ticker].sort_values('Timestamp').reset_index(drop=True)
                ticker_news = news_data[news_data['Ticker'] == ticker].reset_index(drop=True)
                
                # If specific ticker news exists, align it; otherwise default to neutral values
                if not ticker_news.empty:
                    # Map latest news impact onto stock rows
                    avg_impact = ticker_news[sentiment_cols].mean()
                    for col in sentiment_cols:
                        stock_df[col] = avg_impact[col]
                else:
                    for col in sentiment_cols:
                        stock_df[col] = 0.33 if col in ['positive', 'negative', 'neutral'] else 0.0
                
                stock_data[ticker] = stock_df
                print(f"Merged sentiment features into {ticker} dataset (Shape: {stock_df.shape})")

else:
    print("No news data or Transformers unavailable. Adding neutral sentiment defaults...")
    sentiment_cols = ['positive', 'negative', 'neutral', 'polarity', 'strength', 'impact']
    for ticker in TARGET_COLUMNS:
        if ticker in stock_data:
            for col in sentiment_cols:
                stock_data[ticker][col] = 0.33 if col in ['positive', 'negative', 'neutral'] else 0.0
```

```python
# Step 3: Denoise Data for Highest SNR across all 7 stocks
print("\nStep 3: Denoising data...")

def calculate_snr(original, denoised):
    noise = original - denoised
    signal_power = np.mean(original ** 2)
    noise_power = np.var(noise)
    return 10 * np.log10(signal_power / noise_power) if noise_power > 0 else float('inf')

def optimize_denoising(trial, data):
    methods = []
    
    # 1. Wavelet Denoising
    if PYWT_AVAILABLE:
        try:
            wavelet = trial.suggest_categorical('wavelet', ['db1', 'db2', 'sym2'])
            level = trial.suggest_int('wavelet_level', 1, 3)
            coeffs = pywt.wavedec(data, wavelet, level=level)
            threshold = np.std(coeffs[-1]) * np.sqrt(2 * np.log(len(data)))
            denoised_coeffs = [pywt.threshold(c, threshold, mode='soft') for c in coeffs]
            wavelet_denoised = pywt.waverec(denoised_coeffs, wavelet)[:len(data)]
            snr_wavelet = calculate_snr(data, wavelet_denoised)
            methods.append(('wavelet', snr_wavelet, wavelet_denoised))
        except Exception:
            pass

    # 2. Moving Average Denoising
    ma_window = trial.suggest_int('ma_window', 3, 10)
    ma_denoised = pd.Series(data).rolling(window=ma_window, min_periods=1).mean().values
    snr_ma = calculate_snr(data, ma_denoised)
    methods.append(('ma', snr_ma, ma_denoised))

    # 3. Kalman Filter Denoising
    if PYKALMAN_AVAILABLE:
        try:
            kf = KalmanFilter(transition_matrices=[1], observation_matrices=[1], initial_state_mean=data[0])
            kalman_denoised, _ = kf.filter(data)
            kalman_denoised = kalman_denoised.flatten()
            snr_kalman = calculate_snr(data, kalman_denoised)
            methods.append(('kalman', snr_kalman, kalman_denoised))
        except Exception:
            pass

    best_method = max(methods, key=lambda x: x[1] if not np.isinf(x[1]) else -float('inf'))
    return best_method[1], best_method[2]

price_scalers = {}
stock_denoised_data = {}

sentiment_cols = ['positive', 'negative', 'neutral', 'polarity', 'strength', 'impact']

for ticker in TARGET_COLUMNS:
    if ticker in stock_data and not stock_data[ticker].empty:
        print(f"\n--- Processing Denoising for {ticker} ---")
        df_ticker = stock_data[ticker].copy()
        
        # Identify the main price column
        price_col = ticker if ticker in df_ticker.columns else ('Close' if 'Close' in df_ticker.columns else 'Adj Close')
        
        # Scale price data
        scaler = MinMaxScaler()
        scaled_prices = scaler.fit_transform(df_ticker[[price_col]]).flatten()
        price_scalers[ticker] = scaler
        
        # Optimize Denoising using Optuna
        if OPTUNA_AVAILABLE:
            optuna.logging.set_verbosity(optuna.logging.WARNING)
            study_denoise = optuna.create_study(direction='maximize', sampler=TPESampler())
            study_denoise.optimize(lambda trial: optimize_denoising(trial, scaled_prices)[0], n_trials=5)
            _, best_denoised = optimize_denoising(optuna.trial.FixedTrial(study_denoise.best_params), scaled_prices)
            print(f"Best SNR for {ticker}: {calculate_snr(scaled_prices, best_denoised):.2f} dB")
        else:
            best_denoised = pd.Series(scaled_prices).rolling(window=5, min_periods=1).mean().values
            print(f"Default SNR for {ticker}: {calculate_snr(scaled_prices, best_denoised):.2f} dB")
            
        # Store clean data with target price and sentiment columns
        df_denoised = pd.DataFrame({'Scaled_Close': scaled_prices, 'Denoised_Close': best_denoised})
        for col in sentiment_cols:
            if col in df_ticker.columns:
                df_denoised[col] = df_ticker[col].values
            else:
                df_denoised[col] = 0.33 if col in ['positive', 'negative', 'neutral'] else 0.0
                
        stock_denoised_data[ticker] = df_denoised

print("\nStep 3 Complete: All 7 stocks denoised successfully!")
```

```python
# Step 4: Prepare Sequences across all 7 stocks
print("\nStep 4: Preparing sequences...")

SEQ_LENGTH = 30   # Look back 30 days to predict the next day
WINDOW_SIZE = 300 # Keep last 300 days for evaluation/testing

stock_X_deep = {}
stock_X_ml = {}
stock_y = {}
last_300_scaled = {}

for ticker in TARGET_COLUMNS:
    if ticker in stock_denoised_data:
        df_denoised = stock_denoised_data[ticker]
        
        # Grab all numerical feature columns (Scaled Close, Denoised Close, Sentiment scores)
        feature_cols = df_denoised.columns.tolist()
        data_values = df_denoised.values
        
        # Target variable to predict: Denoised Close Price
        target_values = df_denoised['Denoised_Close'].values
        
        # Create rolling 30-day input sequences (X) and future target prices (y)
        X_seq = np.array([data_values[i:i + SEQ_LENGTH] for i in range(len(data_values) - SEQ_LENGTH)])
        y_seq = target_values[SEQ_LENGTH:]
        
        # Store for both Deep Learning (LSTM/GRU) and standard Machine Learning models
        stock_X_deep[ticker] = X_seq
        stock_X_ml[ticker] = X_seq
        stock_y[ticker] = y_seq
        
        # Save last 300 days window for final evaluation
        last_300_scaled[ticker] = data_values[-WINDOW_SIZE:] if len(data_values) >= WINDOW_SIZE else data_values
        
        print(f"[{ticker}] Prepared sequences -> Input shape: {X_seq.shape}, Target shape: {y_seq.shape}")

print("\nStep 4 Complete: Sequences created for all 7 tickers!")
```

```python
# Step 5: Define Model Architecture Factories for Optuna
print("\nStep 5: Defining model optimization factories...")

# Dynamically set feature dimension from the actual data structure
first_ticker = list(stock_denoised_data.keys())[0]
feature_dim = stock_denoised_data[first_ticker].shape[1]
print(f"Detected Feature Dimension: {feature_dim} features per time step.")

def optimize_cnn_lstm(trial, feature_dim):
    filters = trial.suggest_int('filters', 32, 64)
    lstm_units = trial.suggest_int('lstm_units', 50, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    model = Sequential([
        Input(shape=(SEQ_LENGTH, feature_dim)),
        Conv1D(filters=filters, kernel_size=3, padding='same', activation='relu', kernel_regularizer=l2(0.02)),
        LSTM(lstm_units, return_sequences=True, kernel_regularizer=l2(0.02)),
        Dropout(dropout_rate),
        LSTM(lstm_units // 2, kernel_regularizer=l2(0.02)),
        Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_conv1d_lstm(trial, feature_dim):
    filters = trial.suggest_int('filters', 32, 64)
    lstm_units = trial.suggest_int('lstm_units', 50, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    model = Sequential([
        Input(shape=(SEQ_LENGTH, feature_dim)),
        Conv1D(filters=filters, kernel_size=3, padding='same', activation='relu', kernel_regularizer=l2(0.02)),
        LSTM(lstm_units, kernel_regularizer=l2(0.02)),
        Dropout(dropout_rate),
        Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_bilstm(trial, feature_dim):
    lstm_units = trial.suggest_int('lstm_units', 50, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    model = Sequential([
        Input(shape=(SEQ_LENGTH, feature_dim)),
        Bidirectional(LSTM(lstm_units, kernel_regularizer=l2(0.02))),
        Dropout(dropout_rate),
        Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_stacked_lstm(trial, feature_dim):
    lstm_units = trial.suggest_int('lstm_units', 50, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    model = Sequential([
        Input(shape=(SEQ_LENGTH, feature_dim)),
        LSTM(lstm_units, return_sequences=True, kernel_regularizer=l2(0.02)),
        Dropout(dropout_rate),
        LSTM(lstm_units // 2, kernel_regularizer=l2(0.02)),
        Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_gru(trial, feature_dim):
    gru_units = trial.suggest_int('gru_units', 50, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    model = Sequential([
        Input(shape=(SEQ_LENGTH, feature_dim)),
        GRU(gru_units, kernel_regularizer=l2(0.02)),
        Dropout(dropout_rate),
        Dense(1)
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_lightgbm(trial):
    params = {
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'learning_rate': trial.suggest_float('learning_rate', 1e-4, 1e-1, log=True),
        'n_estimators': trial.suggest_int('n_estimators', 100, 500),
        'min_child_weight': trial.suggest_float('min_child_weight', 1e-5, 1.0),
        'verbosity': -1
    }
    return lgb.LGBMRegressor(**params, force_col_wise=True)

def optimize_xgboost(trial):
    params = {
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'learning_rate': trial.suggest_float('learning_rate', 1e-4, 1e-1, log=True),
        'n_estimators': trial.suggest_int('n_estimators', 100, 500),
        'reg_lambda': trial.suggest_float('reg_lambda', 0.1, 10.0)
    }
    return xgb.XGBRegressor(**params)

def optimize_transformer(trial, feature_dim):
    num_heads = trial.suggest_categorical('num_heads', [2, 4, 8])
    ff_dim = trial.suggest_int('ff_dim', 64, 128)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    
    inputs = Input(shape=(SEQ_LENGTH, feature_dim))
    # Match key_dim properly to feature_dim
    x = MultiHeadAttention(num_heads=num_heads, key_dim=max(1, feature_dim // num_heads))(inputs, inputs)
    x = Dropout(dropout_rate)(x)
    x = GlobalAveragePooling1D()(x)
    x = Dense(ff_dim, activation='relu', kernel_regularizer=l2(0.02))(x)
    x = Dropout(dropout_rate)(x)
    outputs = Dense(1)(x)
    
    model = Model(inputs, outputs)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

def optimize_tcn(trial, feature_dim):
    filters = trial.suggest_int('filters', 32, 64)
    kernel_size = trial.suggest_int('kernel_size', 2, 5)
    dropout_rate = trial.suggest_float('dropout_rate', 0.3, 0.6)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 5e-3)
    
    inputs = Input(shape=(SEQ_LENGTH, feature_dim))
    x = Conv1D(filters=filters, kernel_size=kernel_size, padding='same', activation='relu', kernel_regularizer=l2(0.02))(inputs)
    x = Dropout(dropout_rate)(x)
    x = Conv1D(filters=filters, kernel_size=kernel_size, padding='same', activation='relu', kernel_regularizer=l2(0.02))(x)
    shortcut = Conv1D(filters=filters, kernel_size=1, padding='same')(inputs)
    
    x = tf.keras.layers.Add()([shortcut, x])
    x = tf.keras.layers.Activation('relu')(x)
    x = Flatten()(x)
    x = Dense(32, activation='relu', kernel_regularizer=l2(0.02))(x)
    outputs = Dense(1)(x)
    
    model = Model(inputs, outputs)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate), loss='mse')
    return model

print("Step 5 Complete: Model definition functions ready.")
```

```python
models = [
    ('cnn_lstm', optimize_cnn_lstm, 'deep'),
    ('conv1d_lstm', optimize_conv1d_lstm, 'deep'),
    ('bilstm', optimize_bilstm, 'deep'),
    ('stacked_lstm', optimize_stacked_lstm, 'deep'),
    ('gru', optimize_gru, 'deep'),
    ('lightgbm', optimize_lightgbm, 'ml'),
    ('xgboost', optimize_xgboost, 'ml'),
    ('transformer', optimize_transformer, 'deep'),
    ('tcn', optimize_tcn, 'deep')
]

# Step 6: Train and Evaluate Models with Walk-Forward Validation
print("\nStep 6: Training and evaluating models with Walk-Forward Validation...")

def walk_forward_validation(model_fn, model_name, model_type, X, y, price_scaler, trial=None, step_size=100, max_steps=5):
    best_adj_r2 = -float('inf')
    best_model = None
    adj_r2_history = []
    
    n_samples = len(X)
    n_steps = min(max_steps, max(1, (n_samples - SEQ_LENGTH) // step_size))
    
    if n_steps < 1:
        return {'adj_r2': -1.0, 'best_model': None}, []

    n_predictors = feature_dim
    
    # Safely handle sequence vs non-sequence dimensions for ML models
    if model_type == 'ml':
        if len(X.shape) == 3:
            X_proc = X.reshape(X.shape[0], -1)
        else:
            X_proc = X
    else:
        X_proc = X

    for i in range(1, n_steps + 1):
        train_end = int(n_samples * (0.5 + 0.1 * i))  # Expanding window
        if train_end >= n_samples:
            break
            
        test_size = min(step_size, n_samples - train_end)
        if test_size <= 0:
            break

        train_X, train_y = X_proc[:train_end], y[:train_end]
        test_X, test_y = X_proc[train_end:train_end + test_size], y[train_end:train_end + test_size]

        if len(test_X) == 0 or len(test_y) == 0:
            continue

        try:
            # Instantiate model properly (handling both functions and pre-built objects)
            if callable(model_fn):
                if model_type == 'ml':
                    model = model_fn(trial) if trial else model_fn()
                else:
                    model = model_fn(trial, feature_dim) if trial else model_fn(feature_dim)
            else:
                model = model_fn

            # Train model
            if model_type == 'ml':
                model.fit(train_X, train_y)
                preds = model.predict(test_X)
            else:
                early_stop = tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)
                model.fit(train_X, train_y, epochs=8, batch_size=32, validation_split=0.2, verbose=0, callbacks=[early_stop])
                preds = model.predict(test_X, verbose=0).flatten()

            # Reverse scaling for metric calculation
            test_y_inv = price_scaler.inverse_transform(test_y.reshape(-1, 1)).flatten()
            preds_inv = price_scaler.inverse_transform(preds.reshape(-1, 1)).flatten()

            # Adjusted R-squared calculation with safety check
            ss_res = np.sum((test_y_inv - preds_inv) ** 2)
            ss_tot = np.sum((test_y_inv - np.mean(test_y_inv)) ** 2)
            r2 = 1 - (ss_res / ss_tot) if ss_tot != 0 else 0
            n = len(test_y_inv)
            
            if n > (n_predictors + 1):
                adj_r2 = 1 - (1 - r2) * (n - 1) / (n - n_predictors - 1)
            else:
                adj_r2 = r2

            adj_r2_history.append(adj_r2)

            if adj_r2 > best_adj_r2:
                best_adj_r2 = adj_r2
                if model_type == 'deep':
                    best_model = tf.keras.models.clone_model(model)
                    best_model.set_weights(model.get_weights())
                else:
                    best_model = model

        except Exception as e:
            print(f"  [Debug Error in {model_name}]: {e}")
            continue

    return {'adj_r2': best_adj_r2 if best_adj_r2 != -float('inf') else -1.0, 'best_model': best_model}, adj_r2_history


# Master results store across all 7 stocks
stock_model_results = {}

for ticker in TARGET_COLUMNS:
    if ticker in stock_X_deep and ticker in price_scalers:
        print(f"\n================ Running WFV Optimization for {ticker} ================")
        stock_model_results[ticker] = {}
        scaler = price_scalers[ticker]

        X_deep = stock_X_deep[ticker]
        X_ml = stock_X_ml[ticker]
        y_val = stock_y[ticker]

        for name, opt_fn, m_type in models:
            X_curr = X_deep if m_type == 'deep' else X_ml

            if OPTUNA_AVAILABLE:
                optuna.logging.set_verbosity(optuna.logging.WARNING)
                
                def objective(trial):
                    res, _ = walk_forward_validation(opt_fn, name, m_type, X_curr, y_val, scaler, trial=trial, max_steps=3)
                    return res['adj_r2']

                study = optuna.create_study(direction='maximize', sampler=TPESampler())
                study.optimize(objective, n_trials=3)

                best_trial = study.best_trial
                
                # Create final model builder closure with optimized hyperparams
                if m_type == 'deep':
                    final_builder = lambda dim=feature_dim: opt_fn(optuna.trial.FixedTrial(best_trial.params), dim)
                else:
                    final_builder = lambda: opt_fn(optuna.trial.FixedTrial(best_trial.params))
            else:
                final_builder = opt_fn

            # Evaluate best configuration
            wfv_res, history = walk_forward_validation(
                final_builder, name, m_type, X_curr, y_val, scaler, max_steps=5
            )
            
            stock_model_results[ticker][name] = wfv_res
            print(f"[{ticker}] Model: {name:<15} | Best Adj R²: {wfv_res['adj_r2']:.4f}")

print("\nStep 6 Complete: Walk-Forward Validation completed for all models and stocks!")
```

```python
# Step 7: Train and Evaluate Master Pipelines across all 7 stocks
print("\nStep 7: Executing final model evaluation and optimization pipeline...")

stock_results = {}

# Use TARGET_COLUMNS to iterate over your 7 specific stocks
for stock_name in TARGET_COLUMNS:
    if stock_name not in stock_X_deep or stock_name not in price_scalers:
        continue

    print(f"\n================ Processing Final Pipeline for {stock_name} ================")
    results = {}
    
    # Extract stock-specific data and price scaler
    scaler = price_scalers[stock_name]
    y_data = stock_y[stock_name]

    for model_name, model_fn, data_type in models:
        # Check availability flags
        if not (TF_AVAILABLE and data_type == 'deep') and \
           not (LGB_AVAILABLE and model_name == 'lightgbm') and \
           not (XGB_AVAILABLE and model_name == 'xgboost'):
            continue

        print(f"Training {model_name:<15} for {stock_name}...")
        X_data = stock_X_deep[stock_name] if data_type == 'deep' else stock_X_ml[stock_name]

        try:
            if OPTUNA_AVAILABLE:
                optuna.logging.set_verbosity(optuna.logging.WARNING)
                
                def objective(trial):
                    n_samples = len(X_data)
                    train_size = int(0.8 * n_samples)
                    
                    train_X, train_y = X_data[:train_size], y_data[:train_size]
                    val_X, val_y = X_data[train_size:], y_data[train_size:]

                    # Safely handle sequence dimensions for ML vs Deep models
                    if data_type == 'ml':
                        train_X_proc = train_X.reshape(train_X.shape[0], -1) if len(train_X.shape) == 3 else train_X
                        val_X_proc = val_X.reshape(val_X.shape[0], -1) if len(val_X.shape) == 3 else val_X
                        
                        model = model_fn(trial)
                        model.fit(train_X_proc, train_y)
                        preds = model.predict(val_X_proc)
                    else:
                        model = model_fn(trial, feature_dim)
                        early_stop = tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)
                        model.fit(train_X, train_y, epochs=5, batch_size=64, validation_split=0.2, verbose=0, callbacks=[early_stop])
                        preds = model.predict(val_X, verbose=0).flatten()

                    # Reverse scaling using stock-specific scaler
                    val_y_inv = scaler.inverse_transform(val_y.reshape(-1, 1)).flatten()
                    preds_inv = scaler.inverse_transform(preds.reshape(-1, 1)).flatten()

                    # Adjusted R² evaluation
                    ss_res = np.sum((val_y_inv - preds_inv) ** 2)
                    ss_tot = np.sum((val_y_inv - np.mean(val_y_inv)) ** 2)
                    r2 = 1 - (ss_res / ss_tot) if ss_tot != 0 else 0
                    
                    n = len(val_y_inv)
                    adj_r2 = 1 - (1 - r2) * (n - 1) / (n - feature_dim - 1) if n > (feature_dim + 1) else r2
                    
                    # Optuna minimizes objective, so return negative Adj R²
                    return -adj_r2

                study = optuna.create_study(direction='minimize', sampler=TPESampler())
                study.optimize(objective, n_trials=3)

                # Instantiate best trial closure
                if data_type == 'deep':
                    final_builder = lambda dim=feature_dim: model_fn(optuna.trial.FixedTrial(study.best_params), dim)
                else:
                    final_builder = lambda: model_fn(optuna.trial.FixedTrial(study.best_params))
            else:
                final_builder = model_fn

            # Evaluate with fixed walk-forward function
            result, adj_r2_history = walk_forward_validation(
                final_builder, model_name, data_type, X_data, y_data, scaler
            )
            
            results[model_name] = result
            print(f" -> {model_name:<15} | {stock_name} | Best Adj R²: {result['adj_r2']:.6f}")

        except Exception as e:
            print(f"  [Error in {model_name} for {stock_name}]: {e}")
            results[model_name] = {'adj_r2': -1.0, 'best_model': None}

    stock_results[stock_name] = results

print("\nStep 7 Complete: Final model training and evaluation finished!")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

print("\nStep 6.5: Executing Overfitting Analysis Across Models & Stocks...")

# Verify active session state before execution
if 'stock_results' not in globals() or len(stock_results) == 0:
    print("Warning: 'stock_results' is not currently in RAM.")
    print("If your kernel restarted, reconnect your draft session or load your saved data structures.")
else:
    overfitting_records = []

    for stock_name, results in stock_results.items():
        if stock_name not in price_scalers:
            continue
            
        scaler = price_scalers[stock_name]
        y_data = stock_y[stock_name]

        print(f"\n================ Overfitting Check for {stock_name} ================")

        for model_name, result in results.items():
            best_model = result.get('best_model', None)
            if best_model is None:
                continue

            try:
                # Determine feature type and split indices
                is_deep = model_name not in ['xgboost', 'lightgbm']
                X_data = stock_X_deep[stock_name] if is_deep else stock_X_ml[stock_name]

                n_samples = len(X_data)
                train_size = int(0.8 * n_samples)

                train_X, val_X = X_data[:train_size], X_data[train_size:]
                train_y, val_y = y_data[:train_size], y_data[train_size:]

                # Generate train and val predictions
                if not is_deep:
                    train_X_proc = train_X.reshape(train_X.shape[0], -1) if len(train_X.shape) == 3 else train_X
                    val_X_proc = val_X.reshape(val_X.shape[0], -1) if len(val_X.shape) == 3 else val_X

                    train_pred = best_model.predict(train_X_proc)
                    val_pred = best_model.predict(val_X_proc)
                else:
                    train_pred = best_model.predict(train_X, verbose=0).flatten()
                    val_pred = best_model.predict(val_X, verbose=0).flatten()

                # Reverse scaling using stock-specific scaler
                train_y_inv = scaler.inverse_transform(train_y.reshape(-1, 1)).flatten()
                val_y_inv = scaler.inverse_transform(val_y.reshape(-1, 1)).flatten()
                train_pred_inv = scaler.inverse_transform(train_pred.reshape(-1, 1)).flatten()
                val_pred_inv = scaler.inverse_transform(val_pred.reshape(-1, 1)).flatten()

                # Adjusted R² metric calculation
                def compute_adj_r2(y_true, y_pred):
                    ss_res = np.sum((y_true - y_pred) ** 2)
                    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
                    r2 = 1 - (ss_res / ss_tot) if ss_tot != 0 else 0
                    n = len(y_true)
                    return 1 - (1 - r2) * (n - 1) / (n - feature_dim - 1) if n > (feature_dim + 1) else r2

                train_adj_r2 = compute_adj_r2(train_y_inv, train_pred_inv)
                val_adj_r2 = compute_adj_r2(val_y_inv, val_pred_inv)
                overfitting_gap = train_adj_r2 - val_adj_r2

                # Store metrics in dictionary
                stock_results[stock_name][model_name]['train_adj_r2'] = train_adj_r2
                stock_results[stock_name][model_name]['val_adj_r2'] = val_adj_r2
                stock_results[stock_name][model_name]['overfitting_gap'] = overfitting_gap

                # Determine Risk Level
                if overfitting_gap > 0.20:
                    status = "Warning: High Overfitting"
                elif overfitting_gap > 0.10:
                    status = "Caution: Moderate Risk"
                else:
                    status = "Status: Low Risk"

                print(f"Model: {model_name:<15} | Train R²: {train_adj_r2:.4f} | Val R²: {val_adj_r2:.4f} | Gap: {overfitting_gap:.4f} | {status}")

                overfitting_records.append({
                    'Ticker': stock_name,
                    'Model': model_name,
                    'Train_Adj_R2': train_adj_r2,
                    'Val_Adj_R2': val_adj_r2,
                    'Overfitting_Gap': overfitting_gap,
                    'Status': status
                })

            except Exception as e:
                print(f"Error analyzing overfitting for {model_name} on {stock_name}: {e}")

    # Display final overfitting report summary
    if overfitting_records:
        df_overfit = pd.DataFrame(overfitting_records)
        df_overfit.to_csv("overfitting_analysis.csv", index=False)
        print("\nOverfitting summary successfully saved to 'overfitting_analysis.csv'")
```
